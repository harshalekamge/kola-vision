# Kola — train the UI element detector on Colab (free GPU)

Everything heavy (PyTorch, CUDA, ultralytics) downloads on Google's servers, not your internet.

1. Runtime → Change runtime type → **T4 GPU**
2. On your PC, right-click the project's `dataset` folder → Compress to ZIP → `dataset.zip`
3. Run the cells in order. Upload `dataset.zip`, `train.py` and `kola/classes.py` when asked.
4. The last cell downloads `kola_yolo.pt`, `kola_yolo.onnx` and `kola_results.zip` (training plots). Put the models in the project's `models/` folder.
